```python
import glob
import math
import os
import cv2
import numpy as np

FOLDER = "dataset/class_01_watch"
image_paths = sorted(glob.glob(os.path.join(FOLDER, "*.jpg")))[:50]

thumb_size = (120, 120)  # thumbnail size
cols = 10
rows = math.ceil(len(image_paths) / cols)

grid_rows = []
for r in range(rows):
    row_imgs = []
    for c in range(cols):
        idx = r * cols + c
        if idx < len(image_paths):
            img = cv2.imread(image_paths[idx])
            img = cv2.resize(img, thumb_size)
        else:
            img = np.zeros(
                (thumb_size[1], thumb_size[0], 3), dtype=np.uint8
            )  # blank filler
        row_imgs.append(img)
    grid_rows.append(np.hstack(row_imgs))

montage = np.vstack(grid_rows)
cv2.imwrite("perfume_50_preview.jpg", montage)
print("Preview grid saved as 'perfume_50_preview.jpg'!")
```
